```python
# Simplified setup for non-Vocareum environment
import re
from pyspark.sql import functions as F

print("="*80)
print("Starting Demo Environment Setup (Simplified)")
print(f"Using config: {config_path}")
print("="*80)

# Get current user
user = spark.sql("SELECT current_user() as user").collect()[0]["user"]
username = user.split("@")[0]
print(f"\n[0] User: {user}")

# Detect available catalog
try:
    current_catalog = spark.sql("SELECT current_catalog() as cat").collect()[0]["cat"]
    catalog_name = current_catalog
    print(f"  Using current catalog: {catalog_name}")
except:
    # Fallback: get first available catalog
    catalogs = spark.sql("SHOW CATALOGS").collect()
    if catalogs:
        catalog_name = catalogs[0]["catalog"]
        print(f"  Using first available catalog: {catalog_name}")
    else:
        raise Exception("No catalogs available in workspace")

schema_name = f"{username}_agent_demo".replace(".", "_").replace("-", "_").lower()

print(f"\n[1] Setting up catalog and schema...")
print(f"  Catalog : {catalog_name}")
print(f"  Schema  : {schema_name}")

# Create schema if it doesn't exist
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_name}")
spark.sql(f"USE {catalog_name}.{schema_name}")

print(f"\n[2] Schema ready: {catalog_name}.{schema_name}")

# Create environment dictionary
env = {
    "catalog_name": catalog_name,
    "schema_name": schema_name,
    "username": user
}

print("\n" + "="*80)
print("Demo Environment Setup Complete")
print("="*80)
print(f"\nCatalog: {catalog_name}")
print(f"Schema: {schema_name}")
print("\nNote: This is a simplified setup. Full Vocareum setup includes:")
print("  - UC volumes, sample datasets, SQL tools, and PDF generation")
print("  - You may need to create these manually for complete functionality")
print("="*80)

catalog_name = env["catalog_name"]
schema_name = env["schema_name"]
```
